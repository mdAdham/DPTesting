```python
import ipywidgets as widgets
from IPython.display import display

label = widgets.Label("Welcome to Python Widgets")
display(label)

button = widgets.Button(description="Click me")
def clickEvent(e):
    print("Button Clicked")
    
button.on_click(clickEvent)
display(button)

text = widgets.Text(description="Text Random")
display(text)

slider = widgets.IntSlider(value=10, min=0, max=20, description="Age")
display(slider)

check = widgets.Checkbox(value=False, description="Adult")
display(check)

drop = widgets.Dropdown(
    options=["Java", "Python"]
)
display(drop)
```
